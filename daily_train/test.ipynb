```python
from PIL import Image
import os

list_names = []

for filename in os.listdir('.'): 
    if '.png' in filename:
        fig_name = filename
        fig_name = fig_name.strip('.png')
        list_names.append(fig_name)

def crop_img(fig_name): 
    img = Image.open(f'{fig_name}.png')
    crop_area = (75,100,625,340)
    cropped_img = img.crop(crop_area)
    new_name = f'cropped_images\\{fig_name}_cropped.png'
    cropped_img.save(new_name)

for name in list_names: 
    crop_img(name)       
```
